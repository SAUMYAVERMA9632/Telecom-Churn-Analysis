```python
df["TotalCharges"] = df["TotalCharges"].replace(" ", "0")
df["TotalCharges"] = df["TotalCharges"].astype(float)
df.head(21)
```

```python
def conv(value):
   if value == 1:
       return "yes"
   else:
       return "no"

df['SeniorCitizen'] = df["SeniorCitizen"].apply(conv)
```

```python
df.head(30)
```

```python
ax = sns.countplot( x = 'Churn' , data = df)
ax.bar_label(ax.containers[0])
plt.title (" Count of Customer Churn")
plt.show()
```

```python
plt.figure(figsize = (3,4))
gb = df.groupby("Churn").agg({'Churn':"count"})
gb   
plt.pie(gb['Churn'],labels = gb.index, autopct = "%1.2f%%")
plt.title(" Percentage of Churn Customer" ,fontsize = 10)
plt.show()
```

#We can conclude from the above pie chart that 26.54% of customers have churned out .  

```python
plt.figure(figsize = [4,4])
sns.countplot(x = "gender" , data = df, hue= "Churn")
ax.bar_label(ax.containers[0])
plt.title("Churn By Gender")
plt.show()
```

```python
plt.figure(figsize = (9,4))
sns.histplot(x= "tenure", data = df, bins= 72, hue = "Churn")
plt.show()
```

#cutomers who used our services for longer time have stayed , but people who have used for shorter time have churned. 

```python
plt.figure(figsize=(4,4))
ax = sns.countplot(x = "Contract", data = df, hue ="Churn")
ax.bar_label(ax.containers[0])
plt.title("Count of Customers By Contract")
plt.show()
```

#Customers who have month to month plan are more likely to churn than people who have one year or two year plan 

```python
df.columns.values
```

```python
cols = [
    "PhoneService",
    "MultipleLines",
    "InternetService",
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies"
]

fig, axes = plt.subplots(3, 3, figsize=(15, 12))

for i, col in enumerate(cols):
    sns.countplot(x=col, data=df, hue = "Churn", ax=axes[i//3, i%3])
    axes[i//3, i%3].set_title(col)
    axes[i//3, i%3].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

#Summary
- Most customers use PhoneService, while customers without it are fewer.
- Customers without OnlineSecurity, OnlineBackup, DeviceProtection, and TechSupport show higher churn counts.
- Fiber optic customers have a comparatively high number of churned customers.
- Overall, service-related features show noticeable differences in churn across customer groups.
